```python
import pandas as pd
```

```python
df = pd.read_csv("../data/Dataset.csv")
```

```python
legit = df[df['label'] == 0]
```

```python
legit_www = len(legit[legit['url'].str.contains("www.")])
```

```python
phishing = df[df['label'] == 1]
```

```python
phishing_www = len(phishing[phishing['url'].str.contains("www.")])
```

```python
print(f"Legit with 'www': {legit_www / len(legit) * 100:.2f}%")
```

```python
print(f"Phishing with 'www': {phishing_www / len(phishing) * 100:.2f}%")
```
